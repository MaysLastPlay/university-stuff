## Завдання 1: Виконати теоретичні розрахунки асимптотичної складності алгоритмів сорутвання методами вставки та бульбашки.
Сортування вставкою:

Найгірша асимптотична складність - O(n^2). Масив у зворотному порядку, тому кожен елемент доводиться зсувати через усю відсортовану частину.

Найкраща асимптотична складність - O(n). Масив уже відсортований, відбувається лише одне порівняння на крок без жодних переміщень.

Бульбашка:

Найгірша асимптотична складність - O(n^2). Масив у зворотному порядку, що вимагає повного циклу порівнянь та обмінів для кожного елемента.

Найкраща асимптотична складність - O(n). Масив відсортований, цикл миттєво переривається після першого проходу завдяки перевірці на відсутність обмінів.

# Завдання 2: Реалізувати на Python алгоритм сортування методом бульбашки.

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
            
    return arr

my_array = [64, 34, 25, 12, 22, 11, 90]
print("Невідсортований масив:", my_array)

sorted_array = bubble_sort(my_array)
print("Відсортований масив:", sorted_array)
```

```python
import time
import random
import matplotlib.pyplot as plt

def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

n_values = [5, 10, 50, 100, 500, 1000, 2000, 3000, 4000, 5000, 10000]
t_insert = []
t_bubble = []


for n in n_values:
    original_arr = [random.randint(0, 100000) for _ in range(n)]

    arr_for_bubble = original_arr.copy()
    arr_for_insert = original_arr.copy()
    start_time = time.perf_counter()
    bubble_sort(arr_for_bubble)
    end_time = time.perf_counter()
    bubble_time = end_time - start_time
    t_bubble.append(bubble_time)
    start_time = time.perf_counter()
    insertion_sort(arr_for_insert)
    end_time = time.perf_counter()
    insert_time = end_time - start_time
    t_insert.append(insert_time)
    
    print(f"n = {n:<7} | Вставка: {insert_time:.5f} с | Бульбашка: {bubble_time:.5f} с")
plt.figure(figsize=(12, 7))
plt.plot(n_values, t_insert, label='Сортування вставкою (Insertion)', color='blue', marker='o', linewidth=2)
plt.plot(n_values, t_bubble, label='Сортування бульбашкою (Bubble)', color='red', marker='s', linewidth=2)
plt.title('Емпіричне порівняння часу виконання алгоритмів сортування', fontsize=14)
plt.xlabel('Розмір масиву, n', fontsize=12)
plt.ylabel('Час виконання, секунди', fontsize=12)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```


# Контрольні запитання:
1. $O$-нотація показує час роботи в найгіршому випадку (верхня межа), а $\Omega$-нотація — у найкращому (нижня межа).
2. $O(N)$. Оскільки масив уже відсортований, зсуви не потрібні — виконується лише по одному порівнянню для кожного з $N$ елементів.
3. Це алгоритм, який вирішує задачу за прийнятний час (має поліноміальну складність, наприклад $O(N)$ чи $O(N^2)$) та оптимально використовує пам'ять.
4. Це обсяг вхідних даних (наприклад, кількість елементів масиву $N$), від якого безпосередньо залежить час роботи алгоритму.
5. $O(N^3)$. За правилами асимптотики менші доданки та константи відкидаються, залишається лише член із найбільшим ступенем.
