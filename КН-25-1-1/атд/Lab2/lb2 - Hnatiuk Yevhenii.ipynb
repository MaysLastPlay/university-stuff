# Контрольні запитання:
1. Чому не рекомендується використовувати системний Python у Windows?
Щоб уникнути конфліктів версій та не засмічувати операційну систему пакетами, які можуть порушити роботу інших програм або фонових утиліт.

2. Навіщо створювати окреме середовище для кожного проєкту?
Це створює ізольовані "контейнери". Вони гарантують, що проєкт не зламається через оновлення в іншому проєкті, а сам код можна буде легко перенести та запустити на іншому комп'ютері.

3. У чому різниця між базовим середовищем і проєктним?
Базове середовище (base) — це ядро самого менеджера пакетів (наприклад, Conda), і в ньому краще не працювати, щоб випадково не зламати саму програму. Проєктне — це чисте, ізольоване середовище, яке ви створюєте виключно під конкретне завдання.

Для чого використовується environment.yml?
Це файл-конфігурація (для Conda), у якому записаний точний список усіх встановлених бібліотек та їхніх версій. Він дозволяє іншій людині або вам на іншому ПК відтворити ідентичне середовище однією командою.
