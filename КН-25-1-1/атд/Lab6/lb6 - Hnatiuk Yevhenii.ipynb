 # Завдання 2: написати функцію print_n(), що друкує елементи черги з його початку до номера n включно.

```python
class Queue:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def enqueue(self, item):
        self.items.insert(0, item)

    def dequeue(self):
        return self.items.pop()

    def size(self):
        return len(self.items)

    def print_n(self, n):
        limit = min(n, self.size())
        
        print(f"Перші {limit} елементів черги:")
        for i in range(1, limit + 1):
            print(self.items[-i])

q = Queue()
q.enqueue('A')
q.enqueue('B')
q.enqueue('C')
q.enqueue('D')

q.print_n(3)
```

# Завдання 3: оцінити асимптотичну складність (у середньому і в найгіршому разі) процедур search, insert і delete роботи з чергою.

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.insert(0, item)

    def dequeue(self):
        if not self.items:
            return None
        return self.items.pop()

    def search(self, item):
        for i in range(len(self.items) - 1, -1, -1):
            if self.items[i] == item:
                return True
        return False
```

Insert (Додавання): $O(n)$ Метод insert(0, item) змушує масив зсувати всі наявні елементи на одну позицію вправо. 
Delete (Вилучення): $O(1)$ Метод pop() просто забирає останній елемент зі списку, що не потребує зсуву інших елементів.
Search (Пошук): $O(n)$ Прямого доступу немає, тому алгоритм має перевірити кожен елемент по черзі, поки не знайде потрібний.


# Контрольні запитання:
1. Що таке стек і які операції: Стек — структура даних за принципом LIFO (останній прийшов – перший пішов). Основні операції: push (додати на вершину), pop (видалити з вершини), peek (переглянути вершину).
2. Відмінність між стеком та чергою:Стек працює за принципом LIFO (додаємо і забираємо з одного кінця), а черга — за принципом FIFO (перший прийшов – перший пішов: додаємо в кінець, забираємо з початку).
3. Реалізація (масив vs зв'язний список):Масив (список Python): Легко реалізувати, але може потребувати перерозподілу пам'яті при розширенні.Зв'язний список: Швидке додавання/видалення ($O(1)$), динамічний розмір, але потребує більше пам'яті для зберігання вказівників.
4. Застосування:Стек: Стек викликів функцій (рекурсія), перевірка парності дужок, функція "Скасувати" (Undo) в програмах.Черга: Черга завдань до принтера, буферизація потоків даних, обробка запитів клієнтів на сервері.
