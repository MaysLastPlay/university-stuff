допомогою 𝑧-функції.

```python
import heapq

def get_frequencies(text):
    freq_dict = {}
    for char in text:
        freq_dict[char] = freq_dict.get(char, 0) + 1
    return list(freq_dict.keys()), list(freq_dict.values())

def build_huffman_tree(chars, freqs):
    nodes = []
    counter = 0

    for i in range(len(chars)):
        node = {'freq': freqs[i], 'symbol': chars[i], 'left': None, 'right': None, 'huff': ''}
        heapq.heappush(nodes, (freqs[i], counter, node))
        counter += 1

    while len(nodes) > 1:
        left_freq, _, left_node = heapq.heappop(nodes)
        right_freq, _, right_node = heapq.heappop(nodes)
        
        left_node['huff'] = '0'
        right_node['huff'] = '1'

        parent_node = {
            'freq': left_freq + right_freq,
            'symbol': left_node['symbol'] + right_node['symbol'],
            'left': left_node,
            'right': right_node,
            'huff': ''
        }
        
        heapq.heappush(nodes, (parent_node['freq'], counter, parent_node))
        counter += 1
    return heapq.heappop(nodes)[2]

def generate_codes(node, val='', codes=None):
    if codes is None:
        codes = {}
        
    newVal = val + node['huff']
    
    if node['left']:
        generate_codes(node['left'], newVal, codes)
    if node['right']:
        generate_codes(node['right'], newVal, codes)

    if not node['left'] and not node['right']:
        codes[node['symbol']] = newVal
        
    return codes

def huffman_decode(encoded_text, root):
    decoded_text = ""
    current_node = root
    
    for bit in encoded_text:
        if bit == '0':
            current_node = current_node['left']
        else:
            current_node = current_node['right']

        if current_node['left'] is None and current_node['right'] is None:
            decoded_text += current_node['symbol']
            current_node = root
            
    return decoded_text

if __name__ == "__main__":
    message = "abracadabra"
    print(f"Початкове повідомлення: {message}")

    chars, freqs = get_frequencies(message)
    root = build_huffman_tree(chars, freqs)
    huffman_codes = generate_codes(root)
    
    print(f"Коди Гафмена: {huffman_codes}")
    
    encoded_message = "".join([huffman_codes[char] for char in message])
    print(f"Закодоване повідомлення: {encoded_message}")
    
    decoded_message = huffman_decode(encoded_message, root)
    print(f"Декодоване повідомлення: {decoded_message}")
```
