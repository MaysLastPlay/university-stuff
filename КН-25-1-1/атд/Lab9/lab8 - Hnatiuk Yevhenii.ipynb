допомогою 𝑧-функції.

```python
def naive_match(p, t):
    assert len(p) <= len(t)
    occurrences = []
    for i in range(0, len(t) - len(p) + 1):
        match = True
        for j in range(0, len(p)):
            if t[i+j] != p[j]:
                match = False
                break
        if match:
            occurrences.append(i)
    return occurrences

def z_func(s):
    Z = [len(s)] + [0] * len(s)
    if len(s) <= 1: return Z[:len(s)]
    for i in range(1, len(s)):
        if s[i] == s[i-1]: Z[1] += 1
        else: break
            
    r, l = 0, 0
    if Z[1] > 0:
        r, l = Z[1], 1
        
    for k in range(2, len(s)):
        if k > r:
            for i in range(k, len(s)):
                if s[i] == s[i-k]: Z[k] += 1
                else: break
            if Z[k] > 0:
                r, l = k + Z[k] - 1, k
        else:
            nbeta = r - k + 1
            Zkp = Z[k - l]
            if nbeta > Zkp:
                Z[k] = Zkp
            else:
                nmatch = 0
                for i in range(r + 1, len(s)):
                    if s[i] == s[i-k]: nmatch += 1
                    else: break
                l, r = k, r + nmatch
                Z[k] = r - k + 1
    return Z[:-1]
def zMatch(p, t):
    s = p + "$" + t
    Z = z_func(s)
    occurrences = []
    for i in range(len(p) + 1, len(s)):
        if Z[i] == len(p):
            occurrences.append(i - (len(p) + 1))
    return occurrences

def compress_with_z(s):
    z_vec = z_func(s)
    for i in range(1, len(s)):
        if (i + z_vec[i] == len(s)) and (z_vec[i] % i == 0):
            return s[:i]
    return s

if __name__ == "__main__":
    # Тест пошуку підрядка
    text = "hello world, hello grsrysjat, hello everyone"
    pattern = "hello"
    
    print(f"Текст: '{text}'")
    print(f"Шукаємо: '{pattern}'")
    print(f"Наївний пошук (індекси): {naive_match(pattern, text)}")
    print(f"Пошук через Z-функцію (індекси): {zMatch(pattern, text)}\n")
    
    # Тест стиснення рядка
    str_to_compress = "abcabcabcabc"
    print(f"Рядок для стиснення: '{str_to_compress}'")
    print(f"Стиснутий рядок: '{compress_with_z(str_to_compress)}'")
```

1. Наївний пошук: $O(|t| \cdot |p|)$. Працюють два вкладені цикли. У найгіршому випадку кожен символ шаблону перевіряється для кожної позиції тексту.
2. Стиснення (Z-функція): $O(n)$. Побудова Z-функції виконується за лінійний час $O(n)$, а цикл перевірки масиву на періодичність вимагає ще $O(n)$ простих операцій. Разом — $O(n)$.

# Контрольні запитання:
1. Префікс-функція шукає найдовший префікс, який одночасно є суфіксом поточного підрядка.Z-функція шукає найдовший підрядок, що починається з позиції $i$ і повністю збігається з початком (префіксом) усього рядка.
2. Це масив, де кожен елемент показує кількість символів, що збігаються з початком рядка. Роль: дозволяє дуже швидко (за лінійний час $O(N)$) знаходити підрядки в тексті та оптимально стискати рядки.
